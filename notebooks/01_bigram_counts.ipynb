```python
words = open("../data/names.txt", "r").read().splitlines()

print("Number of names:", len(words))
print("First ten:", words[:10])
print("Shortest:", min(len(w) for w in words), "letters")
print("Longest:", max(len(w) for w in words), "letters")
two_letter = [w for w in words if len(w) == 2]
print(len(two_letter), "names with 2 letters:")
print(two_letter)
```
